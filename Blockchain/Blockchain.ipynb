```python
import hashlib
```

```python
class Coins:

    def __init__(self, previous_block_hash, transaction_list):
        self.previous_block_hash = previous_block_hash
        self.transaction_list = transaction_list

        # construct data string
        self.block_data = '---'.join(transaction_list) + '---' + previous_block_hash
        self.block_hash = hashlib.sha256(self.block_data.encode()).hexdigest() # how hash will be calculated
```

```python
t1 = 'Matt sends 2 coins to Daniel'
t2 = 'Charlie sends 5 coins to Rubby'
t3 = 'Rubby sends 3.4 coins to Monica'
t4 = 'Monica sends 8 coins to Matt'
t5 = 'Vanessa sends 5 coins to Daniel'
t6 = 'Daniel sends 3 coins to Rubby'
```

## Transactions
Creating transactions list, later when those are passed as `transaction_list` argument in `Coins` function, hash will be calculated.

```python
initial_block = Coins('Initial Str', [t1, t2])
second_block = Coins(initial_block.block_hash, [t3, t4])
third_block = Coins(second_block.block_hash, [t5, t6])
```

```python
print(initial_block.block_data)
print(initial_block.block_hash)
```

By executing initial block we can see the description of `t1` and `t2` which is connected by `---` in `Coins` class, as it is first block and there is no previous one the `Initial Str` was added after that hash is calculated, in this case it is:<br>
`ad4ce65963eb95a357f08b3c108ae8468a38c847ff473ab36a2fb687aa7a59ba`<br>
This block will be used to construct next blocks, and its hash will be repeated in the next block as well, if even single character is changed the hash will be completly different --> check last cell.


```python
print(second_block.block_data)
print(second_block.block_hash)
```

As `second_block` takes `initial_block` it starts with its hash.<br>
Previous block hash: `ad4ce65963eb95a357f08b3c108ae8468a38c847ff473ab36a2fb687aa7a59ba`<br>
Current hash: `96f4bce2b2e6e6d88aa309ab60fcad46bf937d139df771bcbce63b0c03db8a5a`

```python
print(third_block.block_data)
print(third_block.block_hash)
```
